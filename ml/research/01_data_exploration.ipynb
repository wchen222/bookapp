```python
numeric_years = pd.to_numeric(book_data['Year-Of-Publication'], errors='coerce')
filtered_df = book_data[(numeric_years < 10) | (numeric_years.isna())]
print(filtered_df.shape)
filtered_df.head()
```

```python
numeric_years = pd.to_numeric(book_data['Year-Of-Publication'], errors='coerce')

new_book_data = book_data[(numeric_years >= 500) & (numeric_years.notna())]
new_book_data.shape
```

```python
# check null values for string columns
target_columns = ['ISBN', 'Book-Title', 'Book-Author']

print("True Null Counts:")
print(book_data[target_columns].isna().sum())
```

```python
broken_rows_mask = book_data['Book-Author'].isna() | book_data['Publisher'].isna()

print(book_data[broken_rows_mask][['ISBN', 'Book-Title', 'Book-Author']])
```

```python
for col in target_columns:
    empty_count = (book_data[col].astype(str).str.strip() == "").sum()
    print(f"{col}: {empty_count} blank rows")
```

```python
# full clean
book_data = pd.read_csv('../../data/Books.csv', low_memory=False)
book_data = book_data.drop(columns=['Image-URL-S', 'Image-URL-M', 'Image-URL-L', "Publisher"])
book_data = book_data.dropna(subset=['Book-Author'])
book_data['Year-Of-Publication'] = pd.to_numeric(book_data['Year-Of-Publication'], errors='coerce')
initial_count = len(book_data)
book_data = book_data[book_data['Year-Of-Publication'].notna() & (book_data['Year-Of-Publication'] > 0)]
book_data['Year-Of-Publication'] = book_data['Year-Of-Publication'].astype(int)
print(book_data.shape)
book_data.head()
book_data.to_csv("../data/Books_clean.csv", index=False)
```
