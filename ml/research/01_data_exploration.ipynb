```python
# full clean
book_data = pd.read_csv('../../data/Books.csv', low_memory=False)
book_data = book_data.drop(columns=['Image-URL-S', 'Image-URL-M', 'Image-URL-L', "Publisher"])
book_data = book_data.dropna(subset=['Book-Author'])
book_data['Year-Of-Publication'] = pd.to_numeric(book_data['Year-Of-Publication'], errors='coerce')
initial_count = len(book_data)
book_data = book_data[book_data['Year-Of-Publication'].notna() & (book_data['Year-Of-Publication'] > 0)]
book_data['Year-Of-Publication'] = book_data['Year-Of-Publication'].astype(int)
print(book_data.shape)
book_data.head()
#book_data.to_csv("../data/books_clean.csv", index=False)
```

### Ratings dataset

```python
import pandas as pd

rating_data = pd.read_csv('../../data/Ratings.csv', low_memory=False)
```

```python
rating_data.shape
```

```python
rating_data.head()
```

```python
rating_data = rating_data.drop(columns=['User-ID'])
rating_data.head()
```

```python
total_rows_with_na = rating_data.isna().any(axis=1).sum()
print(total_rows_with_na)
```

A possible problem is ghost ratings, ratings on books I dropped from books.csv \

```python
#clean_book_df = pd.read_csv('../../data/books_clean.csv', low_memory=False)
clean_book_df = book_data.copy()
ratings_df = rating_data[rating_data['ISBN'].isin(clean_book_df['ISBN'])]
```

```python
ratings_df.shape
```

```python
print(1149780 - 1017121)
```

Looks like we removed an entire 132659 ghost records

```python
ratings_stats = ratings_df.groupby('ISBN')['Book-Rating'].agg(['mean', 'count']).reset_index()
new_book_df = clean_book_df.merge(ratings_stats, on='ISBN', how='left')
new_book_df['mean'] = new_book_df['mean'].fillna(0.0)
new_book_df['count'] = new_book_df['count'].fillna(0).astype(int)
new_book_df.to_csv("../../data/books_clean_v2.csv", index=False)
```
